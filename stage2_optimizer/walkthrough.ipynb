# Optimizer : Constant-capacitance model and virtual gates

  

*Sources:*

[1] W. G. van der Wiel, S. De Franceschi, J. M. Elzerman, T. Fujisawa, S. Tarucha,

L. P. Kouwenhoven, *Electron transport through double quantum dots*,

Rev. Mod. Phys. **75**, 1–22 (2003), [arXiv:cond-mat/0205350](https://arxiv.org/abs/cond-mat/0205350);

  

![Double-dot circuit with the five gates](figures/circuit.png)

We start by proving the following theorem that is at the core of the strategy :

  

**Theorem:**

We denote

$$

\mathbf p = \begin{pmatrix} V_2 \\ V_4 \end{pmatrix} \ (\text{plungers}), \qquad

\mathbf b = \begin{pmatrix} V_1 \\ V_3 \\ V_5 \end{pmatrix} \ (\text{barriers}),

$$

$$

\mathbf C_{gP} = \begin{pmatrix} C_{12} & C_{14} \\ C_{22} & C_{24} \end{pmatrix}

\ (\text{dot–plunger}), \qquad

\mathbf C_{gB} = \begin{pmatrix} C_{11} & C_{13} & C_{15} \\ C_{21} & C_{23} & C_{25} \end{pmatrix}

\ (\text{dot–barrier}).

$$

  

If all the capacitances of the device are constant, then changing the barriers by $\delta\mathbf b$ moves **every** stick of the charge-stability diagram by the **same** vector $\delta\mathbf p = \mathbf D\,\delta\mathbf b$, independently of the stick, of the point on the stick and of the electron numbers $(N_1, N_2)$.

  

**Proof :**

  

## 0. Problem and notation

  

The device has five gates. We split them into the **plungers** (the image axes) and the

$$

\mathbf p = \begin{pmatrix} g_2 \\ g_4 \end{pmatrix}, \qquad

\mathbf b = \begin{pmatrix} b_1 \\ b_2 \\ b_3 \end{pmatrix}

          = \begin{pmatrix} g_1 \\ g_3 \\ g_5 \end{pmatrix},

\qquad \mathbf b^{(0)} = \mathbf 0,\quad \mathbf p^{(0)} = (0.15,\ 0.15)\ \text{V}.

$$

  

**Contrast.** For a stick $i$ with signal $I_i$ on the stick and $I_{\mathrm{bg}}$ on the background,

$$

c_i = \left| I_i - I_{\mathrm{bg}} \right| .

$$

  

## 1. Charges

The charge of dot $i$ is the sum of the charges of the capacitors leaving it,
$Q_i = \sum_{k \ne i} C_{ik}(\varphi_i - \varphi_k)$ (source and drain at $0$):

$$
\underbrace{\begin{pmatrix} Q_1 \\ Q_2 \end{pmatrix}}_{\mathbf Q}
=
\underbrace{\begin{pmatrix} C_1 & -C_m \\ -C_m & C_2 \end{pmatrix}}_{\mathbf C}
\underbrace{\begin{pmatrix} \varphi_1 \\ \varphi_2 \end{pmatrix}}_{\boldsymbol\varphi}
-
\underbrace{\begin{pmatrix}
C_{11} & C_{12} & C_{13} & C_{14} & C_{15} \\
C_{21} & C_{22} & C_{23} & C_{24} & C_{25}
\end{pmatrix}}_{\mathbf C_g}
\underbrace{\begin{pmatrix} V_1 \\ V_2 \\ V_3 \\ V_4 \\ V_5 \end{pmatrix}}_{\mathbf v},
\qquad
\begin{aligned}
C_1 &= C_L + C_m + \textstyle\sum_g C_{1g},\\
C_2 &= C_R + C_m + \textstyle\sum_g C_{2g}.
\end{aligned}
$$

## 2. Potentials

$$
\boldsymbol\varphi = \mathbf C^{-1}\mathbf Q + \boldsymbol\alpha\,\mathbf v,
\qquad
\mathbf C^{-1} = \frac{1}{C_1C_2 - C_m^2}\begin{pmatrix} C_2 & C_m \\ C_m & C_1 \end{pmatrix},
\qquad
\boldsymbol\alpha = \mathbf C^{-1}\mathbf C_g = \begin{pmatrix}
\alpha_{11} & \alpha_{12} & \alpha_{13} & \alpha_{14} & \alpha_{15} \\
\alpha_{21} & \alpha_{22} & \alpha_{23} & \alpha_{24} & \alpha_{25}
\end{pmatrix}.
$$

$\alpha_{ig}$ is the lever arm of gate $g$ on dot $i$; $\boldsymbol\varphi$ is linear in $\mathbf v$
because the capacitances are constant.

## 3. Energy

With $\mathbf Q = -|e|\mathbf N$, dropping the terms independent of $\mathbf N$:

$$
F(\mathbf N) = \tfrac12\,\mathbf N^{\top}\mathbf E\,\mathbf N - \mathbf N\cdot\mathbf u
= \tfrac12E_{C1}N_1^2 + \tfrac12E_{C2}N_2^2 + E_{Cm}N_1N_2 - N_1u_1 - N_2u_2,
$$

$$
\mathbf E = e^2\mathbf C^{-1} = \begin{pmatrix} E_{C1} & E_{Cm} \\ E_{Cm} & E_{C2} \end{pmatrix},
\qquad
\mathbf u = |e|\,\boldsymbol\alpha\,\mathbf v .
$$

The five voltages enter only through $(u_1, u_2)$.

## 4. A stick fixes the detuning

The stick between $(N_1{+}1, N_2)$ and $(N_1, N_2{+}1)$ is where both have the same energy:

$$
F(N_1{+}1, N_2) = F(N_1, N_2{+}1)
\iff
\boxed{\ \varepsilon \equiv u_1 - u_2 = E_{C1}\big(N_1 + \tfrac12\big) - E_{C2}\big(N_2 + \tfrac12\big) + E_{Cm}(N_2 - N_1)\ }
$$

The right-hand side is a constant. Its ends (triple points) fix $u_1$ and $u_2$ as well,
e.g. $u_1 = E_{C1}(N_1 + \tfrac12) + E_{Cm}N_2$: **a point of a stick is a fixed
$\mathbf u^\star$**, independent of the voltages.

## 5. Inversion

Splitting plungers and barriers,

$$
\mathbf u = |e|\big(\boldsymbol\alpha_P\,\mathbf p + \boldsymbol\alpha_B\,\mathbf b\big),
\qquad
\boldsymbol\alpha_P = \begin{pmatrix} \alpha_{12} & \alpha_{14} \\ \alpha_{22} & \alpha_{24} \end{pmatrix},
\qquad
\boldsymbol\alpha_B = \begin{pmatrix} \alpha_{11} & \alpha_{13} & \alpha_{15} \\ \alpha_{21} & \alpha_{23} & \alpha_{25} \end{pmatrix}.
$$

Staying on $\mathbf u^\star$ while the barriers move:
$\boldsymbol\alpha_P\,\delta\mathbf p + \boldsymbol\alpha_B\,\delta\mathbf b = \mathbf 0$. With
$\det\boldsymbol\alpha_P = \alpha_{12}\alpha_{24} - \alpha_{14}\alpha_{22} \ne 0$,

$$
\boldsymbol\alpha_P^{-1} = \frac{1}{\alpha_{12}\alpha_{24} - \alpha_{14}\alpha_{22}}
\begin{pmatrix} \alpha_{24} & -\alpha_{14} \\ -\alpha_{22} & \alpha_{12} \end{pmatrix},
\qquad
\boxed{\ \delta\mathbf p = \mathbf D\,\delta\mathbf b, \qquad
\mathbf D = -\boldsymbol\alpha_P^{-1}\boldsymbol\alpha_B\ }
$$

$\mathbf u^\star$ does not appear: the same vector moves every stick. $\blacksquare$

Since $\boldsymbol\alpha_P = \mathbf C^{-1}\mathbf C_{gP}$ and $\boldsymbol\alpha_B = \mathbf C^{-1}\mathbf C_{gB}$,
$\mathbf C$ cancels: $\mathbf D = -\mathbf C_{gP}^{-1}\mathbf C_{gB}$, i.e. when
$C_{14} = C_{22} = 0$

$$
\mathbf D = \begin{pmatrix}
-\dfrac{C_{11}}{C_{12}} & -\dfrac{C_{13}}{C_{12}} & -\dfrac{C_{15}}{C_{12}} \\[8pt]
-\dfrac{C_{21}}{C_{24}} & -\dfrac{C_{23}}{C_{24}} & -\dfrac{C_{25}}{C_{24}}
\end{pmatrix}.
$$

## 6. The quadratic term

  

The model is linear in the voltages, so it has none. It comes from capacitances that

depend on the voltages, e.g. one dot, a plunger $C_p$ and a barrier

$C_b(V_b) = C_b^0(1 + \kappa V_b)$:

  

$$

C_p\,\delta V_p + C_b^0(1 + \kappa V_b)\,V_b = 0

\quad\Longrightarrow\quad

\delta V_p = -\frac{C_b^0}{C_p}\,V_b - \frac{C_b^0}{C_p}\,\kappa\,V_b^2 .

$$

  

It is not predicted by the circuit: it is **measured**. Translation model (18 unknowns):

$$

\boxed{\ t_x(\mathbf b) = \sum_{k=1}^{3} D_{xk}\,b_k + \lambda \sum_{k=1}^{3}\sum_{l=1}^{3} Q_{x,kl}\,b_k b_l,

\qquad x \in \{g_2, g_4\},\quad Q_{x,kl} = Q_{x,lk}\ }

$$

  

# Part 2 — The optimisation, step by step

The seven steps of the procedure, in their **compressed** form
(`optimize(exp)` of `stage2_optimizer/optimize_vg.py`, operators and virtual gates in
`stage2_optimizer/virtual_gates.py`), run here on the example device
`new_experiment(seed=1007)`. One scene image is planned (the reference); after it the
translation model of Part 1 says where every stick is, so only a small window around each
stick is measured. A scene is measured again only as a fallback, when a registration on
patches fails its quality control or lands at the edge of its search. The algorithm only calls `exp.measure`; the figures reuse the images it
measured (no extra pixel), and `exp.reveal()` appears once, in §16, to check the answer.

```python
import csv
import pathlib
import sys

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import FancyBboxPatch, Rectangle
from scipy.ndimage import correlate, gaussian_filter

ROOT = pathlib.Path.cwd() if (pathlib.Path.cwd() / "csd").exists() else pathlib.Path.cwd().parent
sys.path.insert(0, str(ROOT))

from csd import BARRIERS, new_experiment
from stage2_optimizer import optimize_vg as opt
from stage2_optimizer import plot_cost_quality
from stage2_optimizer import virtual_gates as vgm
from stage2_optimizer.virtual_gates import FILTER_LEN, clean, remove_rows, robust_sigma

SEED = 1007
exp = new_experiment(seed=SEED)
plt.rcParams.update({"figure.dpi": 110, "image.origin": "lower", "axes.titlesize": 10})
BNAME = [f"$g_{g[1]}$" for g in BARRIERS]
SCENE_PX = (
    round((vgm.SCENE + 2 * vgm.SEARCH[0]) / vgm.SCENE_STEP) ** 2
)  # one scene image of the full-scene procedure


def n_px(v):
    "Pixels measured for one visit."
    return v.z.size if v.patches is None else sum(z.size for z, _ in v.patches)


def show_visit(ax, v, vmin, vmax):
    "Draw what a visit measured, each window at its place in the (g2, g4) plane."
    if v.patches is None:
        ax.imshow(v.z, extent=v.win.extent(), cmap="gray", vmin=vmin, vmax=vmax)
        return
    ax.set_facecolor("#efe9dc")
    for z, w in v.patches:
        ax.imshow(z, extent=w.extent(), cmap="gray", vmin=vmin, vmax=vmax)
    lo = np.asarray(v.win.center) - v.win.span / 2 - 0.03
    ax.set(xlim=(lo[0], lo[0] + v.win.span + 0.06), ylim=(lo[1], lo[1] + v.win.span + 0.06))
    ax.set_aspect("equal")


def patch_score(v):
    "Registration score S(u) of a patch visit: the sum of its presence patches."
    wide = v.patches[0][1].span > vgm.PATCH + 1e-9
    R = round((vgm.SEARCH_YOUNG if wide else vgm.PATCH_SEARCH) / vgm.SCENE_STEP)
    c = v.patches[0][0].shape[0] // 2
    S = sum(np.clip(-z / v.sigma, 0, 6)[c - R : c + R + 1, c - R : c + R + 1] for z, _ in v.patches)
    return gaussian_filter(S, 1.0), R * vgm.SCENE_STEP * 1e3
```

## 7. Overview

Step 1 measures the only scene image. Steps 2–3 learn to **track** the sticks (virtual
gates, §5–§6) from patches. Step 4 **explores** the whole barrier cube: patches where the
model is still being registered, thin strips in between. Steps 5–7 **conclude** (candidates,
local refinement, choice). Right: where steps 2–4 measure in barrier space.

```python
steps = [
    ("1  Reference", "b = 0, 0.3 V, 1 mV: the reference scene", "sticks $r_i$"),
    ("2  Linear D", "±8 mV per barrier, 40 mV patches", "6 translations → D"),
    ("3  Curvature", "±120 mV and 3 pair points, 60 mV patches", "D and Q fitted"),
    ("4  Grid", "53 patch points, 136 strip points", "depth table K"),
    ("5  Candidates", "no measurement", "ζ ≥ 3.5 pairs"),
    ("6  Refinement", "40 × 12 mV strips, 1 barrier at a time", "summit b"),
    ("7  Answer", "24 mV contrast images, peak and floor", "(g1..g5)"),
]
fig = plt.figure(figsize=(13, 5))
ax = fig.add_axes([0, 0, 0.55, 1])
ax.axis("off")
for k, (name, meas, out) in enumerate(steps):
    y = 0.9 - k * 0.13
    color = "#c6dbef" if k < 3 else ("#fdd0a2" if k == 3 else "#c7e9c0")
    ax.add_patch(
        FancyBboxPatch((0.02, y - 0.045), 0.3, 0.09, boxstyle="round,pad=0.01", fc=color, ec="k")
    )
    ax.text(0.17, y, name, ha="center", va="center", fontsize=11, weight="bold")
    ax.text(0.36, y + 0.015, meas, va="center", fontsize=9)
    ax.text(0.36, y - 0.022, "→ " + out, va="center", fontsize=9, color="dimgray")
    if k < len(steps) - 1:
        ax.annotate("", (0.17, y - 0.085), (0.17, y - 0.05), arrowprops={"arrowstyle": "->"})
ax.set(xlim=(0, 1), ylim=(0, 1))
ax3 = fig.add_axes([0.58, 0.05, 0.4, 0.9], projection="3d")
g, reg = opt.grid_order(5)
ax3.scatter(*g[~reg].T, s=6, c="tab:orange", alpha=0.5, label="step 4: strips")
ax3.scatter(*g[reg].T, s=22, c="tab:red", label="step 4: registered patches")
eye = np.eye(3)
cal = [s * opt.H_SMALL * eye[k] for k in range(3) for s in (1, -1)]
ax3.scatter(*np.array(cal).T, s=30, c="tab:blue", label="step 2: ±8 mV")
cal = [s * opt.H_CURV * eye[k] for k in range(3) for s in (1, -1)]
cal += [opt.H_CROSS * (eye[k] + eye[j]) for k in range(3) for j in range(k + 1, 3)]
ax3.scatter(*np.array(cal).T, s=30, c="tab:green", label="step 3: ±120 mV, pairs")
ax3.set(xlabel=BNAME[0], ylabel=BNAME[1], zlabel=BNAME[2])
ax3.legend(loc="upper left", fontsize=8);
```

## 8. Image operators

On the reference image (1 mV): each scan line has its own offset → the **row median** is
subtracted; a **matched filter** averages 6 mV along the 45° stick direction (the stick
keeps its amplitude, white noise drops by $\sqrt L$); a light blur gives the cleaned image
$z$. The noise level is the robust $\hat\sigma$ (MAD), insensitive to the few stick pixels;
a stick's **depth** is the maximum of $-z$ in a small box around its predicted position.
The same operators are applied to every window measured later, whatever its size.

```python
run = opt.step1_reference(exp)
vg = run.vg
raw = vg.ref_img
rows = remove_rows(raw)
n = round(FILTER_LEN / vg.ref_win.step)
matched = correlate(rows, np.eye(n) / n, mode="nearest")
z_ref = clean(raw, vg.ref_win.step)
sig = robust_sigma(z_ref)

fig, axes = plt.subplots(1, 4, figsize=(15, 4))
for ax, im, title in zip(
    axes,
    (raw, rows, matched, z_ref),
    ("raw", "row medians removed", f"matched filter ({n} px at 45°)", "cleaned z (+ blur)"),
):
    ax.imshow(im, extent=vg.ref_win.extent(), cmap="gray")
    ax.set(title=title, xlabel="$g_2$ (V)")
axes[0].set_ylabel("$g_4$ (V)")
plt.tight_layout()

fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
axes[0].hist(-z_ref.ravel() / sig, bins=200, color="gray", log=True)
axes[0].axvline(4.5, c="tab:red", label="detection level 4.5 σ")
axes[0].set(
    xlabel="$-z / \\hat\\sigma$",
    ylabel="pixels",
    title=f"cleaned image: noise σ = {sig:.3f}; sticks are the tail",
)
axes[0].legend()
i0 = 2
c, r = np.rint(vg.ref_win.to_px(vg.sticks.r[i0])).astype(int)
crop = z_ref[r - 15 : r + 16, c - 15 : c + 16]
axes[1].imshow(-crop, cmap="magma")
axes[1].add_patch(Rectangle((15 - 4.5, 15 - 4.5), 9, 9, fill=False, ec="cyan", lw=2))
axes[1].set(
    title=f"stick {i0}: depth = max of $-z$ in the cyan box = {-crop[11:20, 11:20].min():.2f}",
    xticks=[],
    yticks=[],
)
plt.tight_layout()
```

its connected components closer than 15 mV are merged (fragments of one stick); each stick
is placed at its depth-weighted barycentre $\mathbf r_i$. These positions, at

## 10. Step 2 — registration on patches and linear coefficients $\mathbf D$

Each barrier at ±8 mV: the sticks move by less than 8 mV, so a **40 mV patch** (2 mV)
centred on each reference position still holds its stick, without any model. The
translation is the same for all sticks (Part 1), so only the 6 deepest sticks are measured.
**Registration**: each patch becomes a bounded *presence* map
$s = \min(\max(-z/\hat\sigma, 0), 6)$; as every patch is centred on its prediction, the
score $S(\mathbf u)$ is the **sum of the patches** — only the true shift puts every stick on
a dip at once. Then $\mathbf D_{:,k} = (\mathbf t^+_k - \mathbf t^-_k)/0.016$ V.

```python
opt.calibrate_patches(run)
cal = vg.visits[:15]
v = cal[0]  # g1 = +8 mV
D_fd = np.array([(cal[2 * k].t - cal[2 * k + 1].t) / (2 * opt.H_SMALL) for k in range(3)]).T

fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))
axes[0].imshow(z_ref, extent=vg.ref_win.extent(), cmap="gray")
for _, w in v.patches:
    e = w.extent()
    axes[0].add_patch(Rectangle((e[0], e[2]), w.span, w.span, fill=False, ec="tab:blue", lw=1.5))
others = np.setdiff1d(np.arange(len(vg.sticks.r)), v.sticks)
axes[0].plot(*vg.sticks.r[others].T, "x", c="tab:orange", label="not measured in steps 2–3")
axes[0].set(
    title=f"$g_1$ = +8 mV: {len(v.patches)} patches = {n_px(v)} px (a scene: {SCENE_PX} px)",
    xlabel="$g_2$ (V)",
    ylabel="$g_4$ (V)",
)
axes[0].legend(loc="upper right", fontsize=8)
S, R = patch_score(v)
im = axes[1].imshow(S, extent=(-R, R, -R, R), cmap="Blues")
axes[1].plot(*(1e3 * (v.t - v.t_pred)), "+", c="tab:red", ms=16, mew=2)
axes[1].set(
    title=f"score S(u) = sum of the patches: t = {np.round(1e3 * v.t, 1)} mV",
    xlabel="shift $g_2$ (mV)",
    ylabel="shift $g_4$ (mV)",
)
plt.colorbar(im, ax=axes[1])
for k, w in enumerate(cal[:6]):
    axes[2].annotate(
        "", 1e3 * w.t, (0, 0), arrowprops={"arrowstyle": "->", "color": f"C{k // 2}", "lw": 2}
    )
    axes[2].text(
        *(1e3 * w.t * 1.15),
        f"{BNAME[k // 2]} {'+' if k % 2 == 0 else '−'}",
        color=f"C{k // 2}",
        ha="center",
    )
axes[2].set(
    xlim=(-10, 10),
    ylim=(-10, 10),
    aspect="equal",
    title="the six measured translations",
    xlabel="$t_{g_2}$ (mV)",
    ylabel="$t_{g_4}$ (mV)",
)
axes[2].axhline(0, c="lightgray", lw=0.8)
axes[2].axvline(0, c="lightgray", lw=0.8)
plt.tight_layout()

fig, axes = plt.subplots(1, len(v.patches) + 1, figsize=(2.3 * (len(v.patches) + 1), 2.6))
for ax, (z, w), i in zip(axes, v.patches, v.sticks):
    h = 1e3 * w.span / 2
    ax.imshow(z, extent=(-h, h, -h, h), cmap="gray")
    ax.plot(0, 0, "+", c="tab:orange", ms=10, mew=1.5)
    ax.plot(*(1e3 * (v.t - v.t_pred)), "o", mfc="none", c="tab:blue", ms=12)
    ax.set(title=f"stick {i}", xticks=[-15, 0, 15], yticks=[])
axes[0].set(yticks=[-15, 0, 15], ylabel="mV from the prediction")
ax = axes[-1]
ax.imshow(D_fd, cmap="Blues_r", vmin=-1, vmax=0, origin="upper")
for (a, b), val in np.ndenumerate(D_fd):
    ax.text(b, a, f"{val:.2f}", ha="center", va="center", color="w" if val < -0.5 else "k")
ax.set(
    xticks=range(3),
    xticklabels=BNAME,
    yticks=range(2),
    yticklabels=["$g_2$", "$g_4$"],
    title="measured D (V/V)",
)
plt.tight_layout()
print("the 6 patches of the first image: + prediction (no shift), ○ registered position")
```

## 11. Step 3 — curvature (±120 mV), cross terms and fit

Six measurements at ±120 mV, then one point per pair of barriers at 100 mV, all **centred
with the current model** and on wider patches (60 mV, search ±20 mV) because the prediction
is still rough; if the shift still lands at the edge of the search, the point is measured
again as a full scene (hatched bar). The gap between the translation predicted by $\mathbf D$ alone and the
measured one is the curvature; the quadratic model (§6) is refitted after every
measurement. After these 15 measurements every coefficient of $\mathbf D$ and $\mathbf Q$
has been seen at least once.

```python
model = vg.model
curv = cal[6:12]
labels = [f"{BNAME[np.flatnonzero(v.b)[0]]}{1e3 * v.b[np.flatnonzero(v.b)[0]]:+.0f}" for v in curv]
fig, axes = plt.subplots(1, 3, figsize=(16, 4.4))
for x, ax in enumerate(axes[:2]):
    ax.plot(labels, [1e3 * (D_fd @ v.b)[x] for v in curv], "s", label="linear prediction $D b$")
    ax.plot(labels, [1e3 * v.t[x] for v in curv], "o", mfc="none", ms=11, label="measured $t$")
    for k, v in enumerate(curv):
        ax.annotate(
            f"{1e3 * (v.t - D_fd @ v.b)[x]:+.1f}",
            (k, 1e3 * v.t[x]),
            xytext=(6, -12),
            textcoords="offset points",
            fontsize=8,
        )
    ax.set(
        title=f"translation along $g_{2 * x + 2}$ (gap = curvature, mV)",
        ylabel="mV",
        xlabel="barrier moved (mV)",
    )
    ax.legend(fontsize=8)
err = np.array([1e3 * np.max(np.abs(v.t - v.t_pred)) for v in cal])
group = np.repeat([0, 1, 2], [6, 6, 3])
names = ["±8 mV (40 mV patch)", "±120 mV (60 mV patch)", "pairs at 100 mV (60 mV patch)"]
scene = np.array([v.patches is None for v in cal])
for gk in range(3):
    axes[2].bar(np.flatnonzero(group == gk) + 1, err[group == gk], color=f"C{gk}", label=names[gk])
axes[2].bar(
    np.flatnonzero(scene) + 1, err[scene], fill=False, hatch="///", label="fell back to a scene"
)
axes[2].axhline(1e3 * 0.8 * vgm.PATCH_SEARCH, ls="--", c="C0", lw=1)
axes[2].axhline(1e3 * 0.8 * vgm.SEARCH_YOUNG, ls="--", c="C1", lw=1)
axes[2].set(
    xlabel="calibration measurement",
    ylabel="|measured − predicted| (mV)",
    title="prediction error before each measurement (dashed: usable search)",
)
axes[2].legend(fontsize=8)
plt.tight_layout()
px_cal = run.budget["2 linear D"] + run.budget["3 curvature Q"]
print("D (fit) =\n", model.D.round(3))
print(
    f"calibration: {len(cal)} measurements, {scene.sum()} of them repeated as a scene, {px_cal} px "
    f"in all, failed tries included (full-scene procedure, 12 scenes: {12 * SCENE_PX} px)"
)
```

## 12. Step 4 — exploring the whole barrier cube

189 points: the $5^3$ grid on $\{-0.4, \dots, 0.4\}^3$ V plus the centre of each of its 64
cells (a narrow peak is then never half a grid step away along every barrier at once).
Every window is centred on $\mathbf r_i + \mathbf t_{\text{pred}}(\mathbf b)$ (**virtual
gates**). Two kinds of points:

**Registered points** (27 inner points, then the 26 corners, edge and face centres of the
cube): a 40 mV patch per stick; the translation is measured and the model refitted.

**Points in between** (72 grid points, 64 cell centres): the model interpolates between
registered points, so a **40 × 12 mV strip** per stick is enough to read its depth; no
registration.

```python
opt.step4_grid_compressed(run)
grid_v = vg.visits[run.grid_visits]
_, registered = opt.grid_order(5)
kind = np.where(registered, 0, 1)
kind[125:] = 2
knames = ["registered: 40 mV patches", "between registered points: strips", "cell centres: strips"]
kcol = ["tab:red", "tab:orange", "tab:green"]
px = np.array([n_px(v) for v in grid_v])

fig = plt.figure(figsize=(13, 4.8))
ax3 = fig.add_subplot(1, 2, 1, projection="3d")
for kk in range(3):
    ax3.scatter(
        *run.grid[kind == kk].T,
        s=26 if kk == 0 else 8,
        c=kcol[kk],
        label=f"{knames[kk]} ({np.sum(kind == kk)})",
    )
ax3.set(xlabel=BNAME[0], ylabel=BNAME[1], zlabel=BNAME[2], title="the 189 points")
ax3.legend(fontsize=8, loc="upper left")
ax = fig.add_subplot(1, 2, 2)
ax.plot(np.cumsum(px) / 1e6, c="tab:blue", label="this run (patches, then strips)")
ax.plot(
    np.arange(1, 126) * SCENE_PX / 1e6,
    c="gray",
    ls="--",
    label="one scene per point (full-scene procedure)",
)
ax.axvline(np.sum(kind == 0), c="lightgray", lw=0.8)
ax.set(
    xlabel="grid point (visit order)", ylabel="cumulated pixels (millions)", title="cost of step 4"
)
ax.legend(fontsize=8)
plt.tight_layout()

reg_i, strip_i = np.flatnonzero(kind == 0), np.flatnonzero(kind == 1)
show = [reg_i[1], reg_i[-1], strip_i[np.argmax(np.linalg.norm(run.grid[strip_i], axis=1))]]
zs = np.concatenate([z.ravel() for m in show for z, _ in (grid_v[m].patches or [])])
vmin, vmax = zs.min(), np.percentile(zs, 99.5)
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, m in zip(axes, show):
    v = grid_v[m]
    show_visit(ax, v, vmin, vmax)
    ax.set(
        title=f"{'patches' if kind[m] == 0 else 'strips'}: b = {v.b.round(2)} V, "
        f"t = {np.round(1e3 * v.t).astype(int)} mV, {n_px(v)} px",
        xlabel="$g_2$ (V)",
    )
axes[0].set_ylabel("$g_4$ (V)")
plt.tight_layout()

Kg = np.nan_to_num(opt.K_table(run))
lit = []
for sel in (reg_i, np.flatnonzero(kind > 0)):
    m, i0 = np.unravel_index(np.argmax(Kg[sel]), Kg[sel].shape)
    lit.append((sel[m], i0))
fig, axes = plt.subplots(1, 2, figsize=(9, 3.4))
for ax, (m, i0) in zip(axes, lit):
    v = grid_v[m]
    z, w = v.patches[list(v.sticks).index(i0)]
    e = [1e3 * x for x in w.extent()]
    ax.imshow(z, extent=e, cmap="gray")
    p = 1e3 * (vg.sticks.r[i0] + v.t)
    ax.add_patch(Rectangle(p - 4, 8, 8, fill=False, ec="cyan", lw=1.5))
    ax.set(
        title=f"stick {i0} lit, {'patch' if kind[m] == 0 else 'strip'} ({z.size} px): depth "
        f"{Kg[m, i0]:.1f} in the cyan box",
        xlabel="$g_2$ (mV)",
        ylabel="$g_4$ (mV)",
    )
plt.tight_layout()

reg_v = [grid_v[m] for m in reg_i]
err_lin = np.array([1e3 * np.linalg.norm(D_fd @ v.b - v.t) for v in reg_v])
err_mod = np.array([1e3 * np.linalg.norm(v.t_pred - v.t) for v in reg_v])
fig, ax = plt.subplots(figsize=(7.5, 3.8))
m = np.arange(1, len(reg_v) + 1)
ax.semilogy(m, err_lin, ".", c="tab:orange", label="linear model (D of step 2)")
ax.semilogy(m, err_mod, ".", c="tab:blue", label="quadratic model refitted after each point")
ax.axhline(40, ls="--", c="gray", label="half stick spacing (40 mV)")
ax.axhline(4, ls=":", c="cyan", label="half box where the depth is read (4 mV)")
ax.set(xlabel="registered grid point", ylabel="prediction error (mV)")
ax.legend(fontsize=8)
plt.tight_layout()
print(
    f"model error at the registered points: median {np.median(err_mod):.1f} mV, max {err_mod.max():.1f} mV; "
    f"D alone: median {np.median(err_lin):.0f} mV, max {err_lin.max():.0f} mV"
)
print(
    f"step 4: {px.sum() / 1e6:.2f} M px for 189 points "
    f"({np.sum([v.patches is None for v in grid_v])} scene fallbacks); "
    f"full-scene procedure, 125 scenes: {125 * SCENE_PX / 1e6:.2f} M px"
)
```

## 13. Step 5 — candidates

For each stick, floor $f_i$ = median of its depths over the grid, spread $s_i$ = 1.4826 MAD,
score $\zeta_{m,i} = (K_{m,i} - f_i)/s_i$ (dividing by the stick's own floor removes its
width). Pairs are taken by decreasing $\zeta$, at least 0.3 V apart from those already kept,
while $\zeta \geq 3.5$ (at most 8): one candidate per peak seen. No measurement.

```python
cands = opt.step5_candidates(run)
K = opt.K_table(run)
keep = vg.sticks.confirmed
idx = np.flatnonzero(keep)
Kn = K[:, keep] / np.nanmedian(K[:, keep], axis=0)

fig, ax = plt.subplots(figsize=(13, 3.6))
im = ax.imshow(Kn.T, aspect="auto", cmap="Blues", vmin=1, vmax=5, origin="upper")
for c in cands:
    ax.add_patch(
        Rectangle(
            (c.m - 1.5, np.flatnonzero(idx == c.i)[0] - 0.5), 3, 1, fill=False, ec="tab:red", lw=1.8
        )
    )
for edge in (np.sum(kind == 0), 125):
    ax.axvline(edge - 0.5, c="k", lw=0.8)
ax.set(
    yticks=range(len(idx)),
    yticklabels=idx,
    xlabel="grid point (visit order: patches | strips | cell centres)",
    ylabel="stick",
    title="K / floor (red: candidates)",
)
plt.colorbar(im, ax=ax)

fig, axes = plt.subplots(1, 2, figsize=(13, 3.8))
zf = run.zeta[np.isfinite(run.zeta)]
axes[0].hist(zf, bins=80, color="gray", log=True)
axes[0].axvline(3.5, c="tab:red", label="ζ = 3.5")
axes[0].set(xlabel="ζ", ylabel="(grid point, stick) pairs", title="distribution of the scores")
axes[0].legend()
for c in cands:
    axes[1].plot(K[:, c.i], label=f"stick {c.i}")
    axes[1].plot(c.m, K[c.m, c.i], "*", ms=14, c="tab:red")
axes[1].set(
    xlabel="grid point", ylabel="depth", title="depth of the candidate sticks along the grid"
)
axes[1].legend(fontsize=8)
plt.tight_layout()

fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))
zmax = np.nanmax(np.nan_to_num(run.zeta, nan=0), axis=1)
for ax, (a, b) in zip(axes, ((0, 1), (0, 2), (1, 2))):
    ax.scatter(
        run.grid[:, a], run.grid[:, b], s=10 + 6 * np.clip(zmax, 0, 30), c="tab:blue", alpha=0.25
    )
    for c in cands:
        ax.plot(c.b_grid[a], c.b_grid[b], "*", c="tab:red", ms=14)
        ax.annotate(
            f"stick {c.i}",
            (c.b_grid[a], c.b_grid[b]),
            xytext=(5, 5),
            textcoords="offset points",
            fontsize=8,
        )
    ax.set(
        xlabel=BNAME[a], ylabel=BNAME[b], aspect="equal", title="grid (size: best ζ), candidates ★"
    )
plt.tight_layout()
```

## 14. Step 6 — refining each candidate

Only strips that follow the candidate's stick (40 × 12 mV, 2 mV, centred on
$\mathbf r_i + \mathbf t_{\text{pred}}(\mathbf b)$: 120 px each). One barrier at a time,
9 values over ±0.2 V (pass 1, light), ±0.1 V (pass 2) then ±0.05 V (pass 3, dark, each depth
averaged over two strips); the new value is the vertex of the least-squares parabola through
the five points around the deepest one. Then the strips of one sweep, the path followed in
barrier space, and the stick at its summit.

```python
opt.step6_refine(run, fast=True)
shade = ["lightsteelblue", "cornflowerblue", "tab:blue"]
fig, axes = plt.subplots(len(cands), 3, figsize=(13, 2.9 * len(cands)), squeeze=False)
for row, c in zip(axes, cands):
    for p, (k, xs, ys, xk) in enumerate(c.scans):
        ax, col = row[k], shade[p // 3]
        ax.plot(xs, ys, "o", mfc="none", c=col)
        j = int(np.nanargmax(ys))
        if 2 <= j < len(xs) - 2:
            q = np.polyfit(xs[j - 2 : j + 3], ys[j - 2 : j + 3], 2)
            xx = np.linspace(xs[j - 2], xs[j + 2], 40)
            ax.plot(xx, np.polyval(q, xx), c=col)
        ax.axvline(xk, c="tab:red" if p >= 6 else "mistyrose", lw=1.5)
        ax.set(xlabel=f"{BNAME[k]} (V)", title=f"stick {c.i}: {BNAME[k]} → {xk:+.3f} V")
    row[0].set_ylabel("depth")
plt.tight_layout()

c = cands[0]
k, xs, ys, xk = c.scans[0]
fig = plt.figure(figsize=(13, 3.6))
gs = fig.add_gridspec(2, len(xs), height_ratios=[2.2, 1])
ax = fig.add_subplot(gs[0, :])
ax.plot(xs, ys, "o-", c="tab:blue")
ax.axvline(xk, c="tab:red", lw=1.5)
ax.set(
    xlim=(xs[0] - 0.025, xs[-1] + 0.025),
    ylabel="depth",
    title=f"stick {c.i}, pass 1 along {BNAME[k]}: one strip per value, {sum(z.size for z, _ in c.scan_imgs[0])} px in all",
)
lo, hi = min(z.min() for z, _ in c.scan_imgs[0]), max(z.max() for z, _ in c.scan_imgs[0])
for n, (z, w) in enumerate(c.scan_imgs[0]):
    a = fig.add_subplot(gs[1, n])
    a.imshow(z, cmap="gray", vmin=lo, vmax=hi)
    a.set(xticks=[], yticks=[], xlabel=f"{xs[n]:+.2f} V")

fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))
for ci, c in enumerate(cands):
    path = [c.b_grid.copy()]
    for k, xs, ys, xk in c.scans:
        b = path[-1].copy()
        b[k] = xk
        path.append(b)
    path = np.array(path)
    for ax, (a, b) in zip(axes, ((0, 1), (0, 2), (1, 2))):
        ax.plot(path[:, a], path[:, b], ".-", c=f"C{ci}", label=f"stick {c.i}")
        ax.plot(path[0, a], path[0, b], "*", c=f"C{ci}", ms=12)
        ax.plot(path[-1, a], path[-1, b], "o", c=f"C{ci}", ms=8, mfc="none")
for ax, (a, b) in zip(axes, ((0, 1), (0, 2), (1, 2))):
    ax.set(xlabel=BNAME[a], ylabel=BNAME[b], aspect="equal", title="grid point ★ → summit ○")
axes[0].legend(fontsize=8)
plt.tight_layout()

fig, axes = plt.subplots(1, len(cands), figsize=(3.2 * len(cands), 1.8), squeeze=False)
for ax, c in zip(axes[0], cands):
    z, w = c.summit
    ax.imshow(z, extent=[1e3 * e for e in w.extent()], cmap="gray")
    ax.set(title=f"stick {c.i} at its summit", xlabel="$g_2$ (mV)", ylabel="$g_4$ (mV)")
plt.tight_layout()
for c in cands:
    print(f"stick {c.i:2d}: grid point {c.b_grid.round(2)} → refined {c.b.round(3)} V")
print(f"refinement: {run.budget['6 refinement']} px for {len(cands)} candidates")
```

## 15. Step 7 — contrast, gain and answer

Depths are not comparable between sticks (they depend on the width), so each refined
candidate gets its **absolute contrast** $C$: a 24 mV image at 0.5 mV around the stick, row
medians removed, core = pixels whose smoothed value is below half of the minimum, eroded by
one pixel (outlined), contrast = minus the mean of the core, averaged over four images.

Each stick also has its own factor: a bright stick on a lower peak can beat a dim stick on
the highest one. Candidates within 25 % of the best contrast therefore get the **floor
contrast** $C_0$ of the same stick, measured the same way far from its peak (opposite corner
of the cube), and are ranked by the **gain** $C / C_0$, which compares the peaks themselves.
The best candidate gives $\mathbf b^{\text{final}}$; patches there show the sticks lit with
it (its region); the most contrasted one is the answer, centred on it.

```python
final = opt.step7_answer(run, fast=True)
best = max(cands, key=lambda c: c.gain)
tags = [f"{chr(65 + k)}: stick {c.i}" for k, c in enumerate(cands)]
fig, axes = plt.subplots(2, len(cands), figsize=(3.3 * len(cands), 6.6), squeeze=False)
for k, c in enumerate(cands):
    for ax, imgs, what, val in (
        (axes[0, k], c.contrast_imgs, "at its peak: C", c.contrast),
        (axes[1, k], c.floor_imgs, "floor: C₀", c.floor),
    ):
        if imgs:
            img, core, w = imgs[0]
            e = [1e3 * x for x in w.extent()]
            ax.imshow(img, extent=e, cmap="gray")
            ax.contour(core, levels=[0.5], colors="tab:red", extent=e, origin="lower")
            ax.set(title=f"{tags[k]}, {what} = {val:.1f}", xlabel="$g_2$ (mV)")
        else:
            ax.axis("off")
            why = "same stick, floor above" if np.isfinite(c.floor) else "not shortlisted"
            ax.text(0.5, 0.5, f"{tags[k]}\n{why}", ha="center", va="center")
plt.tight_layout()

fig, axes = plt.subplots(1, 3, figsize=(13, 3.4))
for ax, vals, name in zip(
    axes,
    ([c.contrast for c in cands], [c.floor for c in cands], [c.gain for c in cands]),
    ("absolute contrast C", "floor contrast C₀ of the same stick", "gain C / C₀ (ranking)"),
):
    ax.bar(tags, np.nan_to_num(vals), color=["tab:red" if c is best else "tab:gray" for c in cands])
    ax.set(title=name)
    ax.tick_params(axis="x", labelsize=7)
plt.tight_layout()

v = final["visit"]
g = final["gates"]
zs = np.concatenate([z.ravel() for z, _ in v.patches]) if v.patches else v.z.ravel()
vmin, vmax = zs.min(), np.percentile(zs, 99.5)
fig, axes = plt.subplots(1, 2, figsize=(12, 5.2))
axes[0].imshow(z_ref, extent=vg.ref_win.extent(), cmap="gray", vmin=vmin, vmax=vmax)
axes[0].set(title="start: b = 0 (cleaned)", xlabel="$g_2$ (V)", ylabel="$g_4$ (V)")
show_visit(axes[1], v, vmin, vmax)
for i, C in final["lit"]:
    p = vg.sticks.r[i] + v.t
    axes[1].annotate(
        f"stick {i}: C = {C:.1f}", p, xytext=(14, -22), textcoords="offset points", color="tab:blue"
    )
axes[1].add_patch(plt.Circle((g["g2"], g["g4"]), 0.016, fill=False, color="tab:red", lw=2))
axes[1].set(
    title=f"b_final = {final['b'].round(3)} V (patches) — answer circled", xlabel="$g_2$ (V)"
)
plt.tight_layout()
print("answer:", {k: round(x, 4) for k, x in final["gates"].items()})
```
